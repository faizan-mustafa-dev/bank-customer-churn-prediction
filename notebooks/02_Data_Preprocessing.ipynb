```python
import pandas as pd
```

```python
path = '../data/raw/Churn_Modelling.csv'
dataset = pd.read_csv(path)
```

```python
dataset.head(3)
```

```python
dataset = dataset.drop(columns = ['RowNumber', 'CustomerId', 'Surname'])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(drop = 'first', sparse_output = False)
```

```python
categorical_columns = ['Geography', 'Gender']
```

```python
encoded_array = ohe.fit_transform(dataset[categorical_columns])
```

```python
encoded_df = pd.DataFrame(encoded_array, columns = ohe.get_feature_names_out())
```

```python
dataset = dataset.drop(columns = categorical_columns)
```

```python
dataset = pd.concat([encoded_df, dataset], axis = 1)
```

```python
dataset.to_csv("../data/processed/cleaned_data.csv", index=False)
```
