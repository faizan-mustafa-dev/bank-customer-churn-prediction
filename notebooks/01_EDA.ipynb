```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = '../data/raw/Churn_Modelling.csv'
dataset = pd.read_csv(path)
```

```python
dataset.head(5)
```

```python
dataset.isnull().sum()
```

```python
dataset.describe()
```

```python
dataset.info()
```

```python
plt.figure(figsize = (4,3))
sns.countplot(x = 'Exited', data = dataset)
plt.title("Churn Distribution")
plt.savefig("../reports/churn_distribution.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
dataset_2 = dataset.drop(columns = ['RowNumber', 'CustomerId', 'Surname'])
```

```python
dataset_2.head(3)
```

```python
sns.countplot(x = 'Geography',hue = 'Exited', data = dataset_2)
plt.grid(axis = 'y')
plt.title("Geography vs Churn")
plt.show()
```

```python
sns.boxplot(x = 'Exited', y = 'Age', data = dataset_2)
plt.title("Age vs Churn")
plt.show()
```

```python
plt.figure(figsize = (10,6))
sns.heatmap(dataset_2.corr(numeric_only = True), annot = True)
plt.title("Correlation Heatmap")
plt.show()
```
